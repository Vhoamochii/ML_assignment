# 4. Model Evaluation and Comparison
## Detailed Analysis and Visualization

```python
import sys
sys.path.append('..')

import joblib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.evaluation import ModelEvaluator
from src.utils import load_model

%matplotlib inline
```

## Load Data and Models

```python
# Load test data
X_test = joblib.load('../data/processed/X_test_tfidf.pkl')
y_test = joblib.load('../data/processed/y_test.pkl')

# Load models
models = {
    'Naive Bayes': load_model('../models/naive_bayes.pkl'),
    'Logistic Regression': load_model('../models/logistic_regression.pkl'),
    'SVM': load_model('../models/svm.pkl'),
    'Random Forest': load_model('../models/random_forest.pkl')
}
```

## Load Results

```python
results_df = pd.read_csv('../results/comparison.csv')
print(results_df)
```

## Model Comparison Visualization

```python
evaluator = ModelEvaluator()
evaluator.plot_model_comparison(results_df, save_path='../results/figures/model_comparison.png')
```

## Confusion Matrices

```python
# Plot confusion matrix for best model
best_model_name = results_df.loc[results_df['accuracy'].idxmax(), 'model_name']
best_model = models[best_model_name]

y_pred_best = best_model.predict(X_test)

print(f"Best Model: {best_model_name}")
evaluator.plot_confusion_matrix(y_test, y_pred_best, 
                                save_path=f'../results/figures/confusion_matrix_{best_model_name.replace(" ", "_").lower()}.png')
```

## Detailed Classification Report

```python
print(f"\nDetailed Report for {best_model_name}:")
print("="*60)
evaluator.print_classification_report(y_test, y_pred_best)
```

## Training Time Comparison

```python
plt.figure(figsize=(10, 6))
plt.bar(results_df['model_name'], results_df['train_time'])
plt.xlabel('Model')
plt.ylabel('Training Time (seconds)')
plt.title('Model Training Time Comparison')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../results/figures/training_time.png', dpi=300)
plt.show()
```

## Summary Statistics

```python
print("\nFinal Summary:")
print("="*60)
print(f"Best Model: {best_model_name}")
print(f"Best Accuracy: {results_df['accuracy'].max():.4f}")
print(f"Best F1-Score: {results_df['f1_macro'].max():.4f}")
print(f"\nAll models trained and evaluated successfully!")
```
